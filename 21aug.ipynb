```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
view = pd.read_csv("data.csv")
view
```

```python
import numpy as np

model = LinearRegression()
model.fit(view[["Hours_Student"]], view["Eaxm_Score"])

y_pred = model.predict(view[["Hours_Student"]])
mse = mean_squared_error(view["Eaxm_Score"], y_pred)
rmse = np.sqrt(mse)

print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
```

```python
import matplotlib.pyplot as plt

plt.scatter(view["Hours_Student"], view["Eaxm_Score"], color='blue', label='Actual Data')
plt.plot(view["Hours_Student"], y_pred, color='red', label='Regression Line')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.title('Linear Regression: Hours Studied vs Exam Score')
plt.legend()
plt.show()
```
