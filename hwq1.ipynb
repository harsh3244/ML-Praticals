```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
view = pd.read_csv("datahwq1.csv")
view
```

```python
import numpy as np

model = LinearRegression()
model.fit(view[["Temp(C)"]], view["Ice_Cream_Sales(Units)"])

y_pred = model.predict(view[["Temp(C)"]])
mse = mean_squared_error(view["Ice_Cream_Sales(Units)"], y_pred)
rmse = np.sqrt(mse)

print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
```

```python
import matplotlib.pyplot as plt

plt.scatter(view["Temp(C)"], view["Ice_Cream_Sales(Units)"], color='blue', label='Actual Data')
plt.plot(view["Temp(C)"], y_pred, color='red', label='Regression Line')
plt.xlabel('Temperature (C)')
plt.ylabel('Ice Cream Sales (Units)')
plt.title('Linear Regression: Temperature vs Ice Cream Sales')
plt.legend()
plt.show()
```
