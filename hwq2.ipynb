```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
view = pd.read_csv("datahwq2.csv")
view
```

```python
import numpy as np

model = LinearRegression()
model.fit(view[["Advertising_Sale"]], view["Sales(Units)"])

y_pred = model.predict(view[["Advertising_Sale"]])
mse = mean_squared_error(view["Sales(Units)"], y_pred)
rmse = np.sqrt(mse)

print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
```

```python
import matplotlib.pyplot as plt

plt.scatter(view["Advertising_Sale"], view["Sales(Units)"], color='blue', label='Actual Data')
plt.plot(view["Advertising_Sale"], y_pred, color='red', label='Regression Line')
plt.xlabel('Advertising Spend')
plt.ylabel('Sales (Units)')
plt.title('Linear Regression: Advertising Spend vs Sales')
plt.legend()
plt.show()
```
